## Data Cleaning 02

In this section we will be doing briefly common/basic practices on our second dataframe to prepare our data into more readable insights using pandas.

*Library*

```python
import pandas as pd
```

**Formatting the columns**

```python
df2 = pd.read_csv('../data/raw/ebd_DO_relApr-2026.txt', sep='\t', low_memory = False)

df2['NUMBER OBSERVERS'] = pd.to_numeric(df2['NUMBER OBSERVERS'], errors='coerce').astype('Int64')

df2['LAST EDITED DATE'] = pd.to_datetime(df2['LAST EDITED DATE'], errors='coerce')
df2['OBSERVATION DATE'] = pd.to_datetime(df2['OBSERVATION DATE'], errors='coerce')

columns = ['SUBSPECIES COMMON NAME', 'SUBSPECIES SCIENTIFIC NAME']

df2[columns] = df2[columns].fillna('Unknown').astype(str)
```

```python
df2
```

**Dropping NA values**

We want to drop rows that have missing values in the following columns: 'LATITUDE', 'LONGITUDE', 'OBSERVATION COUNT', 'OBSERVATION DATE', 'NUMBER OBSERVERS', and 'APPROVED'.

```python
df2_nona2 = df2.dropna( subset=['LATITUDE', 
'LONGITUDE', 
'OBSERVATION COUNT', 
'OBSERVATION DATE', 
'NUMBER OBSERVERS', 
'APPROVED'])
```

```python
df2_nona2
```

**Eliminating duplicates and unnecessary columns**

```python
df2_dup3 = df2_nona2.drop_duplicates()

df2_dup4 = df2_dup3.drop(columns=['Unnamed: 52'])
```

We will be using parquet to keep our formatting.

```python
df2_dup4.to_parquet('../data/processed/april-2026-DOchecklist.parquet', index = False)

df2_clean1 = pd.read_parquet('../data/processed/april-2026-DOchecklist.parquet')
```
