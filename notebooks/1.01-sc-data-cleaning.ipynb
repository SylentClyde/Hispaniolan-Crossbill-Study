## Data Cleaning 01

In this section we will be doing briefly common/basic practices on our first dataframe to prepare our data into more readable insights using pandas.

*Library*

```python
import pandas as pd
```

**Formatting the columns**

```python
df1 = pd.read_csv('../data/raw/(Loxia_Megapaglaga)ebd_hiscro_relApr-2026.txt', sep='\t', low_memory = False)

df1['NUMBER OBSERVERS'] = pd.to_numeric(df1['NUMBER OBSERVERS'], errors='coerce').astype('Int64')

df1['LAST EDITED DATE'] = pd.to_datetime(df1['LAST EDITED DATE'], errors='coerce')
df1['OBSERVATION DATE'] = pd.to_datetime(df1['OBSERVATION DATE'], errors='coerce')

columns = ['SUBSPECIES COMMON NAME', 'SUBSPECIES SCIENTIFIC NAME']

df1[columns] = df1[columns].fillna('Unknown').astype(str)
```

```python
df1
```

**Dropping NA values**

We want to drop rows that have missing values in the following columns: 'LATITUDE', 'LONGITUDE', 'OBSERVATION COUNT', 'OBSERVATION DATE', 'NUMBER OBSERVERS', and 'APPROVED'.

```python
df1_nona2 = df1.dropna( subset=['LATITUDE', 
'LONGITUDE', 
'OBSERVATION COUNT', 
'OBSERVATION DATE', 
'NUMBER OBSERVERS', 
'APPROVED'])
```

```python
df1_nona2
```

**Eliminating duplicates and unnecessary columns**

```python
df1_dup3 = df1_nona2.drop_duplicates()

df1_dup4 = df1_dup3.drop(columns=['Unnamed: 52'])
```

We will be using parquet to keep our formatting.

```python
df1_dup4.to_parquet('../data/processed/crossbill_clean.parquet', index = False)

df1_clean1 = pd.read_parquet('../data/processed/crossbill_clean.parquet')
```
